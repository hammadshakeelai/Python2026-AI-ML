```python
import pandas as pd
import re
from pathlib import Path

# File paths and labels
file_config = {
    'f.txt': 'Office/Project Ohook Activation',
    'r.txt': 'Windows KMS Activation',
    'v.txt': 'Windows Z-Method Activation',
    't.txt': 'Main Complete Script'
}

# Pattern to match: UUID_KEY_%f%VERSION_EDITION
full_pattern = r'([a-f0-9]{8}-[a-f0-9]{4}-[a-f0-9]{4}-[a-f0-9]{4}-[a-f0-9]{12})_([A-Z0-9\-]+)_+(.+?)(?=\n|$)'

all_keys = []

for filename, description in file_config.items():
    try:
        with open(filename, 'r', encoding='utf-8', errors='ignore') as f:
            content = f.read()
            
            # Find all UUID_Key_Version patterns
            matches = re.findall(full_pattern, content)
            for uuid, key, version_info in matches:
                if key and not key.startswith('%'):
                    # Clean version info
                    version_info = version_info.replace('%f%', '').strip()
                    
                    all_keys.append({
                        'File': filename,
                        'Type': description,
                        'UUID': uuid,
                        'Key': key,
                        'Version/Edition': version_info,
                        'Company/Purpose': ''  # You can fill this in
                    })
    except Exception as e:
        print(f"Error reading {filename}: {e}")

# Create dataframe
df = pd.DataFrame(all_keys)
print(f"Found {len(df)} keys\n")
print(df.head(20))
```
